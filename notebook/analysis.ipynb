```python
## What is the total outstanding principal (₹) of all assessable active loans? Outstanding = principal − principal repayments.
R=pd.Timestamp('2026-10-15')
rate_asof=gold_rates[gold_rates['rate_date']<=R].sort_values('rate_date')['rate_24k_per_gram'].iloc[-1]
```

```python
repaid=repayment[repayment['payment_type']=='principal'].groupby('loan_id')['amount'].sum()
loans_new.head()
```

```python
book=loans_new[(loans_new['status']=='active')  & (loans_new['principal']>0) & loans_new['gold_weight_gm'].notna()]
book['outstanding']=book['principal']-book['loan_id'].map(repaid).fillna(0)
book['collateral']=book['gold_weight_gm']*book['purity']/24 *rate_asof
```

```python
book['current_ltv']=book['outstanding']/book['collateral']
```
